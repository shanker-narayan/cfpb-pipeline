```python
dbutils.widgets.text("catalog", "cfpb_dev")
catalog = dbutils.widgets.get("catalog")

spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog}")
for schema in ["bronze", "silver", "gold", "reference"]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}")
```

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS cfpb_pipeline
URL 's3://shanker-cfpb-pipeline/'
WITH (STORAGE CREDENTIAL cfpb_storage_cred)
COMMENT 'Storage for CFPB pipeline';
```

```python
catalog = dbutils.widgets.get("catalog")

spark.sql(f"""
  CREATE EXTERNAL VOLUME IF NOT EXISTS {catalog}.bronze.landing
  LOCATION 's3://shanker-cfpb-pipeline/{catalog}/landing'
  COMMENT 'Raw CFPB API files before COPY INTO'
""")

dbutils.fs.put(f"/Volumes/{catalog}/bronze/landing/_test.txt", "hello", True)
display(dbutils.fs.ls(f"/Volumes/{catalog}/bronze/landing/"))
```
